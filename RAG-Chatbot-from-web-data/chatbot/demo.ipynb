```python
# Import chatbot functions

from utils import store_docs, get_response, get_chroma_client
```

```python
# List of all regional tourism websites
target_urls = [
    "https://visitethiopia.et",
    "https://visitoromia.org",
    "https://visitamhara.travel",
    "https://tourismtigrai.com",
    "https://visitsidama.travel",
    "https://visitsouthethiopia.et"
]

# Check existing data to avoid duplicates
db = get_chroma_client()

try:
    existing_data = db.get()
    existing_urls = {m['url'] for m in existing_data['metadatas'] if 'url' in m}
    print(f"📊 Found {len(existing_urls)} existing URLs in database")
except Exception:
    existing_urls = set()
    print("📊 No existing data found")

# Scrape and Store
for url in target_urls:
    if url in existing_urls:
        print(f"⏩ Skipping {url} - already indexed.")
        continue

    print(f"🌐 Indexing regional site: {url}...")
    try:
        stored = store_docs(url)
        if stored:
            print(f"✅ Success: {url}")
        else:
            print(f"⚠️ No data indexed for: {url}")
    except Exception as e:
        print(f"❌ Failed to scrape {url}: {e}")

print("✨ All regional knowledge bases are updated and ready.")
```
